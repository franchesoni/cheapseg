```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.signal.windows import gaussian

# Gaussian kernel (std = 10, window size = 100)
gaussian_kernel = gaussian(500, std=200)
gaussian_kernel /= gaussian_kernel.sum()  # Normalize to make sure sum is 1

k1 = np.load('sample_error_rate_k1.npy')
ksqrt = np.load('sample_error_rate_ksqrt.npy')

plt.figure()
plt.plot(k1, label='k=1')
plt.plot(ksqrt, label='k=sqrt(n)')
plt.legend()
plt.xlabel('Number of clicks')
plt.ylabel('Error rate')

# now the same but smoothed 
plt.figure()
plt.plot(np.convolve(k1, gaussian_kernel, mode='valid'), label='k=1')
plt.plot(np.convolve(ksqrt, gaussian_kernel, mode='valid'), label='k=sqrt(n)')
plt.legend()
plt.xlabel('Number of clicks')
plt.ylabel('Error rate')
plt.show()

print(np.mean(k1[:min(len(k1), len(ksqrt))]))
print(np.mean(ksqrt[:min(len(k1), len(ksqrt))]))
```
